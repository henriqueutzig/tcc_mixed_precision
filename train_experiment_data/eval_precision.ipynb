```python
import os
import pandas as pd
from glob import glob

# Directory containing the CSV files
csv_dir = 'grace/results_precision_batch'

# Find all CSV files in the directory
csv_files = glob(os.path.join(csv_dir, '*.csv'))

# Helper function to get base name without suffix
def get_base_name(filename, suffix):
    if filename.endswith(suffix + '.csv'):
        return filename[:-len(suffix + '.csv')]
    elif filename.endswith('.csv'):
        return filename[:-4]
    return filename

amd_raw_metrics = pd.DataFrame()
df_metrics = pd.DataFrame()

# Group files by base name (with and without suffix)
for file_path in csv_files:
    filename = os.path.basename(file_path)
    if filename.endswith('raw_metrics.csv'):
        amd_raw_metrics = pd.concat([amd_raw_metrics, pd.read_csv(file_path)], ignore_index=True)
    else:
        df_metrics = pd.concat([df_metrics, pd.read_csv(file_path)], ignore_index=True)

# Find the index of the row with the maximum throughput for each model and precision
best_idx = df_metrics.groupby(['model', 'precision'])['avg_throughput_samples_per_sec'].idxmax()

# Select the best performing rows
best_performing_df = df_metrics.loc[best_idx]

# Sort by model and precision for better readability
best_performing_df = best_performing_df.sort_values(['model', 'precision'])

# Display the relevant columns
print("Best performing batch size for each model and precision:")
with pd.option_context('display.max_rows', None, 'display.max_columns', None, 'display.width', 1000):
    print(best_performing_df[['model', 'precision', 'batch_size', 'avg_throughput_samples_per_sec', 'peak_memory_mb', 'avg_utilization_percent', 'avg_power_watts']])
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

# Create a figure-level plot (relplot) to easily facet by model
g = sns.relplot(
    data=df_metrics,
    x='batch_size',
    y='avg_throughput_samples_per_sec',
    hue='precision',
    col='model',
    kind='line',
    marker='o',
    col_wrap=3, # Adjust the number of columns in the grid
    facet_kws={'sharey': False, 'sharex': False} # Allow axes to have different scales
)

# Add a title to the entire figure
g.fig.suptitle('Throughput vs. Batch Size for Each Model (L40S)', y=1.03)

# Improve layout
plt.tight_layout()
plt.show()
```

```python
best_performing_df['throughput_per_watt'] = (
    best_performing_df['avg_throughput_samples_per_sec'] / best_performing_df['avg_power_watts']
)
best_performing_df['throughput_per_gb'] = (
    best_performing_df['avg_throughput_samples_per_sec'] / (best_performing_df['peak_memory_mb'] / 1024)
)

# Optional: sort by combined efficiency metric
best_performing_df['efficiency_score'] = (
    0.33 * best_performing_df['avg_throughput_samples_per_sec'].rank(ascending=False) +
    0.33 * best_performing_df['throughput_per_watt'].rank(ascending=False) +
    0.33 * best_performing_df['throughput_per_gb'].rank(ascending=False)
)

best_performing_df = best_performing_df.sort_values(['model', 'precision', 'efficiency_score'], ascending=[True, True, False])

print("Best performing batch size and efficiency metrics:")
with pd.option_context('display.max_columns', None):
    print(best_performing_df[['model', 'precision', 'batch_size',
                              'avg_throughput_samples_per_sec',
                              'throughput_per_watt',
                              'throughput_per_gb',
                              'avg_utilization_percent',
                              'peak_memory_mb',
                              'avg_power_watts']])
```

```python
# Load L40S data (from grace directory)
csv_dir_l40s = 'grace/results_precision_batch'
csv_files_l40s = glob(os.path.join(csv_dir_l40s, '*.csv'))
df_l40s = pd.DataFrame()
for file_path in csv_files_l40s:
    filename = os.path.basename(file_path)
    if not filename.endswith('raw_metrics.csv'):
        df_l40s = pd.concat([df_l40s, pd.read_csv(file_path)], ignore_index=True)

# Add GPU identifier to L40S data
df_l40s['gpu'] = 'L40S'

# Load RTX 4090 data
csv_dir_4090 = '4090/results_precision_batch'
csv_files_4090 = glob(os.path.join(csv_dir_4090, '*.csv'))
df_4090 = pd.DataFrame()
for file_path in csv_files_4090:
    filename = os.path.basename(file_path)
    if not filename.endswith('raw_metrics.csv'):
        df_4090 = pd.concat([df_4090, pd.read_csv(file_path)], ignore_index=True)

# Add GPU identifier to RTX 4090 data
df_4090['gpu'] = 'RTX 4090'

# Load RX 7900 data
csv_dir_7900 = '7900/results_precision_batch'
csv_files_7900 = glob(os.path.join(csv_dir_7900, '*.csv'))
df_7900 = pd.DataFrame()
for file_path in csv_files_7900:
    filename = os.path.basename(file_path)
    if not filename.endswith('raw_metrics.csv'):
        df_7900 = pd.concat([df_7900, pd.read_csv(file_path)], ignore_index=True)
df_7900['gpu'] = 'RX 7900XT'

# Create separate heatmaps for each GPU
gpus_data = [
    (df_l40s, 'L40S'),
    (df_4090, 'RTX 4090'),
    (df_7900, 'RX 7900XT')
]

for df_gpu, gpu_name in gpus_data:
    # Skip if dataframe is empty or has no valid data
    if df_gpu.empty or 'avg_throughput_samples_per_sec' not in df_gpu.columns:
        print(f"Skipping {gpu_name} - no valid data")
        continue
    
    # Filter out rows with NaN values in required columns
    df_gpu_clean = df_gpu.dropna(subset=['model', 'precision', 'avg_throughput_samples_per_sec', 'batch_size'])
    
    if df_gpu_clean.empty:
        print(f"Skipping {gpu_name} - no clean data after filtering")
        continue
    
    # Get the row with maximum throughput for each model-precision combination
    max_throughput_idx = df_gpu_clean.groupby(['model', 'precision'])['avg_throughput_samples_per_sec'].idxmax()
    max_throughput_data = df_gpu_clean.loc[max_throughput_idx]
    
    # Create pivot tables for both throughput and batch size
    throughput_matrix = max_throughput_data.pivot(index='model', columns='precision', values='avg_throughput_samples_per_sec')
    batch_size_matrix = max_throughput_data.pivot(index='model', columns='precision', values='batch_size')
    
    # Create custom annotations combining throughput and batch size
    annotations = pd.DataFrame(index=throughput_matrix.index, columns=throughput_matrix.columns, dtype=str)
    for i in range(len(throughput_matrix.index)):
        for j in range(len(throughput_matrix.columns)):
            if pd.notna(throughput_matrix.iloc[i, j]):
                throughput = throughput_matrix.iloc[i, j]
                batch_size = batch_size_matrix.iloc[i, j]
                annotations.iloc[i, j] = f'{throughput:.1f}\n(bs={int(batch_size)})'
            else:
                annotations.iloc[i, j] = ''
    
    # Create a separate figure for each GPU
    plt.figure(figsize=(8, 6))
    
    # Create heatmap
    sns.heatmap(throughput_matrix, 
                annot=annotations, 
                fmt='', 
                cmap='viridis', 
                cbar_kws={'label': 'Throughput (samples/sec)'},
                annot_kws={'size': 9})
    
    plt.title(f'Maximum Throughput and Optimal Batch Size for {gpu_name}')
    plt.xlabel('Precision')
    plt.ylabel('Model')
    plt.tight_layout()
    plt.show() 
    
    # Create a separate figure for each GPU
    plt.figure(figsize=(8, 6))
```

```python
# Combine all GPU data into one DataFrame
df_combined = pd.concat([df_l40s, df_4090, df_7900], ignore_index=True)

# Create separate plots for each GPU with better visual distinction
for gpu_name in df_combined['gpu'].unique():
    gpu_data = df_combined[df_combined['gpu'] == gpu_name]
    
    # Set the order of precision categories to ensure bf16 is drawn last (on top)
    precision_order = ['fp32', 'fp16', 'bf16']
    gpu_data['precision'] = pd.Categorical(gpu_data['precision'], categories=precision_order, ordered=True)
    gpu_data = gpu_data.sort_values('precision')
    
    # Create a faceted plot for this specific GPU with different markers and styles
    g = sns.relplot(
        data=gpu_data,
        x='batch_size',
        y='avg_throughput_samples_per_sec',
        hue='precision',
        hue_order=precision_order,  # Explicitly set the order
        style='precision',  # Add style mapping for better distinction
        col='model',
        kind='line',
        markers={'bf16': 'o', 'fp16': 's', 'fp32': '^'},  # Different markers
        dashes={'bf16': (2, 2), 'fp16': (1, 0), 'fp32': (5, 2, 1, 2)},  # Different line styles
        # palette='viridis',  # Use same color palette as heatmaps
        col_wrap=3,
        height=4,
        aspect=1.2,
        facet_kws={'sharey': False, 'sharex': False},
        linewidth=2,  # Make lines thicker
        markersize=8  # Make markers larger
    )
    
    # Set titles and labels
    g.fig.suptitle(f'Training Throughput vs. Batch Size by Model and Precision ({gpu_name})', y=1.02)
    g.set_axis_labels("Batch Size", "Throughput (samples/sec)")
    g.set_titles("Model: {col_name}")
    
    # Improve legend
    g._legend.set_title('Precision')
    
    plt.tight_layout()
    plt.show()
```
