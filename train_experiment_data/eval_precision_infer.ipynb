# L40S

```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# The path to the CSV file
file_path = 'grace/results_inference_batch/summary.csv'

# Read the CSV file into a pandas DataFrame
df_summary = pd.read_csv(file_path)

# Display the first few rows of the DataFrame to verify it was read correctly
# For each model and precision combination, find the row with the maximum throughput
# This identifies the most performant batch size based on throughput
most_performant_idx = df_summary.groupby(['model', 'precision'])['throughput'].idxmax()
most_performant_df = df_summary.loc[most_performant_idx]

# Display the resulting DataFrame
print("Most performant configuration for each model and precision (based on max throughput):")
most_performant_df
```

```python
# Set the aesthetic style of the plots
sns.set_style("whitegrid")

# Create a faceted plot to compare throughput across different batch sizes for each model and precision
g = sns.relplot(
    data=df_summary,
    x='batch_size',
    y='throughput',
    hue='precision',
    col='model',
    kind='line',
    marker='o',  # Add markers to indicate the exact batch sizes tested
    col_wrap=3,  # Wrap the columns after 3 plots
    height=4,
    aspect=1.2,
    facet_kws={'sharey': False, 'sharex': False}
)

# Set titles and labels
g.fig.suptitle('Throughput vs. Batch Size by Model and Precision (L40S)', y=1.03)
g.set_axis_labels("Batch Size", "Throughput (samples/sec)")
g.set_titles("Model: {col_name}")

# Adjust layout to make room for the suptitle
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Read data for all three GPUs
df_l40s = pd.read_csv('grace/results_inference_batch/summary.csv')
df_4090 = pd.read_csv('4090/results_inference_batch/summary.csv')
df_7900 = pd.read_csv('7900/results_inference_batch/summary.csv')

# Add GPU identifier to each dataframe
df_l40s['gpu'] = 'L40S'
df_4090['gpu'] = 'RTX 4090'
df_7900['gpu'] = 'RX 7900XT'

# Create separate heatmaps for each GPU
gpus_data = [
    (df_l40s, 'L40S'),
    (df_4090, 'RTX 4090'),
    (df_7900, 'RX 7900XT')
]

for df_gpu, gpu_name in gpus_data:
    # Get the row with maximum throughput for each model-precision combination
    max_throughput_idx = df_gpu.groupby(['model', 'precision'])['throughput'].idxmax()
    max_throughput_data = df_gpu.loc[max_throughput_idx]
    
    # Create pivot tables for both throughput and batch size
    throughput_matrix = max_throughput_data.pivot(index='model', columns='precision', values='throughput')
    batch_size_matrix = max_throughput_data.pivot(index='model', columns='precision', values='batch_size')
    
    # Create custom annotations combining throughput and batch size
    annotations = throughput_matrix.copy().astype(str)
    for i in range(len(throughput_matrix.index)):
        for j in range(len(throughput_matrix.columns)):
            model = throughput_matrix.index[i]
            precision = throughput_matrix.columns[j]
            if pd.notna(throughput_matrix.iloc[i, j]):
                throughput = throughput_matrix.iloc[i, j]
                batch_size = batch_size_matrix.iloc[i, j]
                annotations.iloc[i, j] = f'{throughput:.1f}\n(bs={batch_size})'
    
    # Create a separate figure for each GPU
    plt.figure(figsize=(8, 6))
    
    # Create heatmap
    sns.heatmap(throughput_matrix, 
                annot=annotations, 
                fmt='', 
                cmap='viridis', 
                cbar_kws={'label': 'Throughput (samples/sec)'})
    
    plt.title(f'Inference: Optimal Batch Size for {gpu_name}')
    plt.xlabel('Precision')
    plt.ylabel('Model')
    plt.tight_layout()
    plt.show()
```

```python
# Combine all GPU data into one DataFrame
df_combined = pd.concat([df_l40s, df_4090, df_7900], ignore_index=True)

# Create separate plots for each GPU with better visual distinction
for gpu_name in df_combined['gpu'].unique():
    gpu_data = df_combined[df_combined['gpu'] == gpu_name]
    
    # Set the order of precision categories to ensure bf16 is drawn last (on top)
    precision_order = ['fp32', 'fp16', 'bf16']
    gpu_data['precision'] = pd.Categorical(gpu_data['precision'], categories=precision_order, ordered=True)
    gpu_data = gpu_data.sort_values('precision')
    
    # Create a faceted plot for this specific GPU with different markers and styles
    g = sns.relplot(
        data=gpu_data,
        x='batch_size',
        y='throughput',
        hue='precision',
        hue_order=precision_order,  # Explicitly set the order
        style='precision',  # Add style mapping for better distinction
        col='model',
        kind='line',
        markers={'bf16': 'o', 'fp16': 's', 'fp32': '^'},  # Different markers
        dashes={'bf16': (2, 2), 'fp16': (1, 0), 'fp32': (5, 2, 1, 2)},  # Different line styles
        # palette='viridis',  # Use same color palette as heatmaps
        col_wrap=3,
        height=4,
        aspect=1.2,
        facet_kws={'sharey': False, 'sharex': False},
        linewidth=2,  # Make lines thicker
        markersize=8  # Make markers larger
    )
    
    # Set titles and labels
    g.fig.suptitle(f'Inference: Throughput vs. Batch Size by Model and Precision ({gpu_name})', y=1.02)
    g.set_axis_labels("Batch Size", "Throughput (samples/sec)")
    g.set_titles("Model: {col_name}")
    
    # Improve legend
    g._legend.set_title('Precision')
    
    plt.tight_layout()
    plt.show()
```
